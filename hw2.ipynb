# Самый популярный маршрут по каждому из направлений (Domestic/International)

```python
df1=df[['Код АП', 'Направление']]
df1.groupby(['Направление','Код АП']).value_counts()
```

# Вывод части таблицы с рейсами категории C и маршрутом САЛЕХАРД

```python
df[(df['Катег'] == 'C') & (df['Маршрут'] == 'САЛЕХАРД')]
```

# Вывод на экран результата из пункта выше с сортировкой по убыванию Baggage Weight

```python
df1=df[(df['Катег'] == 'C') & (df['Маршрут'] == 'САЛЕХАРД')]
df1.sort_values(by=['Baggage Weight'], ascending=[False])
```

# Средняя МВМ и Вместимость с группировкой по направлению и типу ВС

```python
df.groupby(['Направление', 'Тип ВС'])[['МВМ, тонн', 'Вместимость']].mean()
```
